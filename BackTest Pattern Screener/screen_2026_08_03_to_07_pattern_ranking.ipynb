# Cached forecast + bullish-pattern ranking: 3–7 August 2026

CPU-only. This notebook does **not** load Kronos, Torch, checkpoints, or Hugging Face. It reuses both saved `all_forecast_paths.parquet` files, anchors the first future return to the actual 31 July close, and applies the historically optimized pattern ranking.

```python
from pathlib import Path
import subprocess
import sys

cwd = Path.cwd().resolve()
repo = cwd.parent if cwd.name == 'BackTest Pattern Screener' else cwd
pattern_dir = repo / 'BackTest Pattern Screener'
assert pattern_dir.exists(), f'Pattern folder not found from {cwd}'
print('Repository:', repo)
```

```python
subprocess.run(
    [sys.executable, '-m', 'pip', 'install', '-q', '-r', str(pattern_dir / 'requirements.txt')],
    check=True,
)
```

```python
subprocess.run(
    [sys.executable, str(pattern_dir / 'run_cached_future_pattern_screen.py')],
    check=True,
)
```

```python
import pandas as pd

output_dir = pattern_dir / 'future_screens' / '2026-08-03_to_07'
consensus = pd.read_csv(output_dir / 'consensus_top30_pattern.csv')
consensus.loc[consensus['consensus_rank'] <= 10].sort_values(['target_date', 'consensus_rank'])
```

```python
for model in ('validated_no_refit_e15', 'production_refit_e4'):
    selected = pd.read_csv(output_dir / f'{model}_top30_pattern.csv')
    print(model)
    display(selected.loc[selected['selected_rank'] <= 10, [
        'target_date', 'selected_rank', 'ticker', 'secondary_score',
        'net_pattern_score', 'final_ranking_score', 'top_bullish_pattern',
        'pattern_support', 'pattern_penalty',
    ]])
```
